## Data Cleaning :

### Standardize columns :
- This prevents code errors in languages like python if column names contaims hidden spaces, trailing whitespaces and special characters.

-Simplifies merging : Joining datasets ( matching key columns)

_Common Practices in standardization_

- use of lowercase
- replace spaces with underscores
- rename columns

```python
sales_dataset.head()
```

```python
# Standardizing the column names
# 1. Stripping any preceding or trailing spaces

sales_dataset.columns = sales_dataset.columns.str.strip()

# 2. Changes the columns to lowercase

sales_dataset.columns = sales_dataset.columns.str.lower()

# 3. Replace empty spaces with underscores 

sales_dataset.columns = sales_dataset.columns.str.replace("","_")

# 4. # Remove % from the discount column and replace with " Percentage"

sales_dataset = sales_dataset.rename(columns=("Discount (%)"))



```

```python
# Function to standardize columns.

sales_dataset.columns = sales_dataset.columns.str.strip().str.lower().str.replace("","_").str.replace("_(%)","_percentage", regex=False)
# Regex : regular expression( a sequence of special characters that form a serch pattern)
```
